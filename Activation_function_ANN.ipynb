 Training Speed (Fast vs Slow)
 Fast Training:

ReLU activation

Because ReLU does not suffer from vanishing gradient problem.

So Model 1 (ReLU – 6 neurons) trained fastest and best.

 Slow Training:

Sigmoid (slowest)

Because sigmoid compresses values between 0 and 1 → gradients become very small.

So Model 2 (Sigmoid) is slowest.

Medium Speed:

Tanh (faster than sigmoid, slower than relu)

Final Conclusion

Fastest + Best Accuracy → ReLU (6 neurons)

Slowest → Sigmoid

Increasing neurons does NOT always improve accuracy.

Activation function matters more than number of neurons.
